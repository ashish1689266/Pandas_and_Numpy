```python
# we will import the data from csv file
# this is data of 50 cities for hourly aqi
import pandas as pd
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
# now to just see data we can see the first rows
df.head()
```

```python
df.tail()
```

```python
# lets describe the data
df.describe()
```

```python
df.info()
```

```python
# now we have our data so we will learn how to access that
# again we have labells and we have columns remember
# so here if you want to access with labels where we have not given any labels to rows they will be same as normal index but columns are different and string names
# we use loc(for label access)
df.loc[0:2, ["city","aqi", "timestamp"]] # first index is for rows, second is for columns, if for column you give numeric index, thats an error

# and a very important thing in loc the ending index is inclusive, means that is included that is why you are getting three rows
```

```python
# now we will see index based access
# for this we use iloc
df.iloc[0:3, 0:3] # so we are giving integer values as no strings allowed proper indexing and ending index is exclusive here as in all indexing
```

```python
# now we do not use loc and iloc for single value. we can use but we won't
# we will use at and iat now they are same as loc and iloc just for sungle values
# so at is for labelled access and iat is for indexed access
df.at[1, "city"] # first row and then column
```

```python
# now for iat
df.iat[1, 2] # only proper indexes are used because no string values are allowed
```

```python
#now again we can use the values and change them but normally it would lead to change is original data so we make a copy of it and then change
cities = df["city"] # this is just giving you a view of original data if you make changes in cities now it will reflect in original data
# and this is risky that is why use this
cities = df["city"].copy()
cities[0] = "Kolkata"
cities[0]
```

```python
# now you can access column directly
df[["city","aqi","temperature"]]  # you have to give multiple columns as a list
```

```python
# here we will learn about filtering of data
# so to filter the data what we can do we directly put a condition inside []

df[df["aqi"] > 100] # this will give you all rows which have aqi greater than aqi
```

```python
# you can give multiple conditions
df[(df["aqi"] > 100) & (df["temperature"] < 30)]
```

```python
# you can give multiple conditions
df[(df["aqi"] > 100) & (df["temperature"] > 30)]
```

```python
# now we will save this data in a new variable and they are giving us full row but we can get only required columns
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30) ][["city", "aqi"]]
aqi_data
```

```python
# now we can use loc and iloc here for data accessing
# but since labells are same as before because pandas preserve it because they can carry a meaning to it
# so for labells you must give labells which are correct like first label is 6 
aqi_data.loc[6, "city"]
```

```python
# with iloc so indexes are not preserved it just creates new one so we can start with zero
aqi_data.iloc[0, 0]
```

```python
# now for this filtering we have something better
# query method
df.query("aqi > 100 & temperature > 30")
```

```python
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
```

```python
# for storing data into new one query by default gives us a copy but we will still use the .copy()
aqi_data = df.query("aqi > 100 & temperature > 30")[["city", "aqi"]].copy()
aqi_data
```
