```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
df
```

```python
# section A
df[["Name", "Sex", "Age", "Fare", "Survived"]]
```

```python
# section B task 1
df[(df["Sex"] == "female")][["PassengerId", "Name"]]
```

```python
df[(df["Fare"] > 30)][["PassengerId", "Name"]]
```

```python
# with both conditions
df[(df["Sex"] == "female") & (df["Fare"] > 30)][["Name", "PassengerId"]]
```

```python
# section c
df.groupby("Pclass").agg(
    Average_fare=("Fare", "mean"),
    Average_age=("Age", "mean"),
    Survival_rate=("Survived", "mean")
)
```
