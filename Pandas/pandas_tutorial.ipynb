```python
dict = {
    "Name": ["Ashish", "Raj","Nikita","Pranjal"],
    "CGPA": [8.48, 8.8, 9.7, 7.6],
    "City": ["Gwalior", "Indore", "Delhi", "Wellington"]
}
```

```python
# now we can load this dictionary data as a dataframe which is a 2d labelled dataset which is a collection of 1d 
# labelled dataset called series

# now to see and have data as dataframe we will use our pandas which we have imported as pd

df = pd.DataFrame(dict) # like this we can create dataframe, it is created in a varuable name df you can print it
print(df)
# after printing you can see how it has arranged it
# keys become columns, each value of columns makes row and the counting on left of it is labelling
# it starts from 0 same as indexing but we can change it
```

```python
# nowwe gonna work on the series in detail

# series is the 1D labelled data structure
# means it is not only indexed but labelled but they are king of same
# index here is positional argument only eg, s[0] will get you first element and so on
# label here is kind of name to it like by default it starts from 0, so s[0] will get you first element and so on 
# but you can change it while making the series and after it suppose you are storing the age and want to name of people as lebel

# normal series

s = pd.Series([23, 45, 22, 55, 33]) # normal way index = label here
print(s, type(s))
print()

# with changed index
s = pd.Series([26, 28, 27, 28, 26], index=["Nikita","Ashish","Raj","Pranjal","Devraj"]) # here we have labelled them with names
# but you can see and that is irony that the name labelled here is index😂😂

print(s, type(s))
print()

# we can print all labelles
print(s.index)
```

```python
# now we see more about series
# pandas is based on numpy so thi series data structure is based on ndarray of numpy
# so it stores homogeneous data means data of same type, and if given non same data, it type cast it
s = pd.Series([1, 2, 3, 4, 5]) # here data is homogeneous
print(s, type(s))

# now lets cahnge it 
s = pd.Series([1, 2, 3, "Hello", 4.6])
print(s, type(s))
```

```python
# ypu can do operations here by vectorization like
s = pd.Series([1, 2, 3, 4, 5])
print(s * 5)
print(s / 2)

# now you can not change the size of Series (it is immutable) but you can change the values of series (they are mutable)
s[0] = 100
print(s)

# now value at index 0 become 100
# but if you drop the value this will not work
s.drop(0) # looks that it worked but no lets print s again
print()
print(s) # you will found 100 again, hence dropping something will result into reduction of size and this is not possible so 
# we store series with dropped element in a new variable and make new series

new_s = s.drop(0)
print(s)
print(new_s)

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
s_sum = s1 + s2
print(s_sum) # these are series for now wedo that for series, if i have more i will share
```

```python
# Now we see dataframes in detail
# they are 2D ndarray so lets make them with dictionary once again
dict = {
    "Name": ["Ashish", "Raj", "Pranjal"],
    "Age": [28, 27, 28],
    "cgpa": [8.4, 8.8, 8.3]
}

df = pd.DataFrame(dict) # creating it with dictionary and here key become column names and values of them become columns
# and each value from column combinely becomes row

print(df, type(df))
print(df.index) # indexes for row
print(df.columns) # printing the column names
```

```python
# now lets make it with list becuase it is 2D so we create a 2D list
list1 = [["Ashish", 28, 8.4],["Raj", 27, 8.8],["Pranjal", 28, 8.3]]
# you can see how this is each individual list is a complete row and these values separately gives us column number
list2 = [["Ashish", "Raj", "Pranjal"],[28, 27, 28],[8.4, 8.8, 8.3]]
df = pd.DataFrame(list2, index=["Name", "Age", "cgpa"]) # here we are giving indexes and in next one we are giving the 
# columns
print(df, type(df))
# when we create dataframe out of this we need to give column heads or names because weithout it would not look nice
df = pd.DataFrame(list1, columns=["Name", "Age", "cgpa"])
print(df, type(df))

# so in this example you can see you can do it both ways
```

```python
# now we will create it with numpy array
import numpy as np
np_arr = np.array([["Ashish", 28, 8.4],["Raj", 27, 8.8],["Pranjla", 28, 8.3]]) # here we know axis 0 gives row for 2D and axis 1 gives column so according to data we follow the each individual list as complete row
print(np_arr)

# now we create dataframe
df = pd.DataFrame(np_arr, columns =["Name", "Age", "CGPA"])
print(df, type(df))
```
