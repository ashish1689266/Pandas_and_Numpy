```python
# importing json library to import jason files
import json
```

```python
# now we are making a function in which open file in read mode and return it
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# now we will create a function to clean data and structure it
def clean_data(data):
    text_to_num = {
        "one": 1, "two": 2, "three": 3, "four": 4, "five": 5
    }
    cleaned_data = []
    unique_user = set()
    
    for user_data in data:
        # here we are dealing with inconsistent data means somewhere values are in different format, so converting them in same format
        raw_rating = str(user_data["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user_data["rating"] = raw_rating

        # now we will seek for missing values and give them a None value so they show something and hold a valid value though it is None
        raw_age = user_data.get("age") # get will give you None if no value found
        if raw_age == None:
            user_data["age"] = None # this is how we change the value to none

        # now we are making clean data in a new list and to do that we are using sets for unique identification
        if user_data["name"].strip() not in unique_user:
            unique_user.add(user_data["name"])
            cleaned_data.append(user_data)
        # we can do this also
        # if user_data["name"].strip() in unique_user:
        #     # continue
        # unique_user.add(user_data["name"])
        # cleaned_data.append(user_data)
    return cleaned_data
```

```python
clean_data = clean_data(data) # calling the function here
for user in clean_data:
    print(user)
```

```python
# now we have to find insights from data
def insights(data):
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    average_rating = total_rating / len(data)
    print(f"Average rating is {average_rating}")

    poor_ratings = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_ratings += 1
    poor_rating_percentage = (poor_ratings/len(data)) * 100
    print(f"Percentage of poor ratings is {poor_rating_percentage}%")
```

```python
insights(clean_data)
```

```python
# now we need a recommendation system which can help us know whether we need to recommend same brand products or different
def recomm(data):
    recommendation = []
    for user in data:
        curr_recomm = {}
        if float(user["rating"]) >= 4:
            curr_recomm["name"] = user["name"]
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["name"] = user["name"]
            curr_recomm["brand"] = "Samsung"
        recommendation.append(curr_recomm)
    return recommendation
```

```python
recommendation = recomm(clean_data)
for user in recommendation:
    print(user)
```
